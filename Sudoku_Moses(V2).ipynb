```python
import random
from collections import defaultdict
import math
from boards_list import boards  #grabs the boards from boards.py
```

```python
def find_empty(board): #checks for if the number is a zero (empty)
  for i in range(len(board)):
    for j in range(len(board[0])):
      if board[i][j] == 0:
        return (i, j)
  return None
```

```python
def valid(board, num, pos): #checks for valid numbers within the row, col, and box
  for i in range(len(board[0])):
    if board[pos[0]][i] == num and pos[1] != i:
      return False
  for j in range(len(board)):
    if board[j][pos[1]] == num and pos[0] != j:
      return False
  box_row = 3*(pos[0]//3)
  box_col = 3*(pos[1]//3)
  box_end_row = box_row + 3
  box_end_col = box_col + 3

  for i in range(box_row, box_end_row):
    for j in range(box_col, box_end_col):
      if board[i][j] == num and (i, j) != pos:
        return False
  return True
```

```python
def solution(board):
  if len(board) != 9: #Checks for if the board has nine rows
    return False
  for row in board:
    if len(row) != 9: #Checks for if the board has nine columns
      return False

  find = find_empty(board) #makes sure board is empty
  if find is not None:
    return False

  for i in range(len(board)):
    for j in range(len(board[0])):
      if not valid(board, board[i][j], (i,j)):
        return False
      if not (1<= board[i][j]<=9):
        return False
  return True
```

# **Backtracking search algorithm**

```python
def backtrack(board):
    find = find_empty(board)
    if find is None: #once board is not empty, returns completed board
        return board
    else:
        row, col = find
    for i in range(1, 10):
      if valid(board, i, (row, col)): #searches board for an empty spot and places a valid number
          board[row][col] = i
          result = backtrack(board)

          if result is not None:
            return result

          board[row][col] = 0 #if current path results in a incorrect board, returns up by replacing the spot with a zero
    return None
```
